### 영어 문서 카운트 벡터 추출

```python
sample_corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]
```

```python
# Vectorizer 객체 생성
from sklearn.feature_extraction.text import CountVectorizer

# 특징 집합 생성
cVectorizer = CountVectorizer()

# 카운트 벡터 생성
dtm = cVectorizer.fit_transform(sample_corpus)


# DTM 출력
print(dtm.toarray())
```

```python
# 참고
feature_set = cVectorizer.vocabulary_
print(feature_set)

# index 0부터 출력
print(sorted(feature_set.items(), key=lambda x : x[1]))
```

### 한글 문서 카운트 벡터 추출

```python
sample_corpus = [
    '자연어처리 강의를 시작하겠습니다.',
    '자연어처리는 재미있습니다.',
    '밥을 먹고 강의를 듣고 있습니다.',
    '이번 자연어처리 강의는 한국어 자연어처리입니다.' 
]
```

```python
from konlpy.tag import Okt
def my_tokenizer(text):
    my_tags = ['Noun']
    my_stopwords = ['이번']
    return [ word for word, tag in Okt().pos(text) if tag in my_tags and word not in my_stopwords]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Vectorizer 객체 생성
vectorizer = CountVectorizer(tokenizer=my_tokenizer)

# 특징 집합 생성
vectorizer.fit(sample_corpus)
vectorizer.get_feature_names_out()

# 카운트 벡터 생성
dtm = vectorizer.transform(sample_corpus)
print(dtm.toarray())
# DTM 출력
```

## TF-IDF 벡터 생성
* sklearn.feature_extraction.text.TfidfVectorizer 사용
1. 한글 토크나이저 정의
2. 특징 추출 모델 생성 : Vectorizer -> fit()
3. 문서별 특징 벡터 추출 : transform()

### 한글 토크나이저 정의

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(tokenizer=my_tokenizer)
vectorizer.fit(sample_corpus)
print(vectorizer.get_feature_names_out())

dtm = vectorizer.transform(sample_corpus)
dtm.toarray()

```

### 특징 추출 모델 생성 : TfIdfVectorizer

```python
# TfIdfVectorizer 객체 생성 (tokenizer 지정, 최대 단어 수지 지정)
from sklearn.feature_extraction.text import TfidfVectorizer
from konlpy.tag import Okt
tVectorizer = TfidfVectorizer(tokenizer=my_tokenizer, max_features=1000)
```

### 문서별 특징 벡터 추출

```python
# 특징 집합과 특징 벡터 계산을 위한 데이터 추출
from sklearn.feature_extraction.text import TfidfVectorizer

tVectorizer = TfidfVectorizer(tokenizer=my_tokenizer, max_features=1000)
```

```python
# 특징 벡터 추출
tfidf_dtm = tVectorizer.fit_transform(sample_corpus)
```
